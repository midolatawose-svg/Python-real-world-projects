```python
branch1_stock = [
[10, 20, 30],
[40, 50, 60]
]
branch2_stock = [
[1, 2, 3],
[4, 5, 6]
]
total_stock = [
[0, 0, 0],
[0, 0, 0]
]
```

```python
for i in range(len(branch1_stock)):
    for j in range(len(branch1_stock[i])):
        total_stock[i][j] = branch1_stock[i][j] + branch2_stock[i][j]
print(total_stock)
```

```python
total_stock = [
[11, 22, 33],
[44, 55, 66]
]
total = 0
```
